```python
import uncertainties
from uncertainties import ufloat
from uncertainties.umath import *
from asymmetric_uncertainty import a_u
import numpy as np
```

```python
def age_0_2(prot, lower_err, upper_err):
    prot_e = a_u(prot,upper_err,lower_err)
    # prot_e = ufloat(prot, (upper_err+lower_err)/2)
    # log_age = ufloat(0.0094,0.0002)*prot_e+ufloat(0.1909,0.0114)
    log_age = a_u(0.0094,0.0002,0.0002)*prot_e+a_u(0.1909,0.0114,0.0114)
    age = 10**log_age
    return age

def age_0_2_segmented(prot, lower_err, upper_err):
    prot_e = a_u(prot,upper_err,lower_err)
    # prot_e = ufloat(prot, (upper_err+lower_err)/2)
    # log_age = ufloat(0.0094,0.0002)*prot_e+ufloat(0.1909,0.0114)
    log_age = a_u(0.0621,0.0024,0.0024)*prot_e-a_u(1.0437,0.0380,0.0380)-a_u(0.0528,0.0025,0.0025)*(prot_e-a_u(23.4933,0.7643,0.7643))
    age = 10**log_age
    return age

def age_2_6(prot, lower_err, upper_err):
    prot_e = a_u(prot,upper_err,lower_err)
    # prot_e = ufloat(prot, (upper_err+lower_err)/2)
    # log_age = ufloat(0.0094,0.0002)*prot_e+ufloat(0.1909,0.0114)
    log_age = a_u(0.0041,0.0001,0.0001)*prot_e+a_u(0.3691,0.0126,0.0126)
    age = 10**log_age
    return age

def age_2_6_segmented(prot, lower_err, upper_err):
    prot_e = a_u(prot,upper_err,lower_err)
    # prot_e = ufloat(prot, (upper_err+lower_err)/2)
    # log_age = ufloat(0.0094,0.0002)*prot_e+ufloat(0.1909,0.0114)
    log_age = a_u(0.0561,0.0012,0.0012)*prot_e-a_u(0.8900,0.0185,0.0185)-a_u(0.0521,0.0012,0.0012)*(prot_e-a_u(24.1888,0.4268,0.4268))
    age = 10**log_age
    return age
```

```python
# From our work's Prot
toi442_age = age_0_2(28.64,1.6,0.9)
print(toi442_age)
toi442_age = age_0_2_segmented(28.64,1.6,0.9)
print(toi442_age)
```
